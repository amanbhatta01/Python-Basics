**Question 2 : Explain the difference between List and Tuple in Python.**

 - A List and a Tuple are both used to store multiple values in Python, but they are different in the following ways:

 1. List is mutable, which means we can change, add, or remove values after creating it.

 2. Tuple is immutable, which means we cannot change its values after creating it.

 3. List is written using square brackets [ ], while Tuple is written using parentheses ( ).

**Question 3 : What is a function in Python? Why are functions useful?**

 - A function in Python is a block of code that performs a specific task. We can create a function once and use it multiple times whenever needed.

 - Functions are useful because:-

 1. They reduce repeated code.

 2. They make the program easy to understand.

 3. They make code easy to reuse and maintain.

**Question 5 : Write a Python program to check whether a number is even or odd using conditional statements.**

CODE:-

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

**Question 7 : Create a list of five numbers and print the maximum number from the list.**

CODE:-

```python
numbers = [10, 25, 7, 40, 15]

maximum = max(numbers)

print("Maximum number is:", maximum)
```

**Question 8 : Write a Python program to remove duplicate values from a list using a set.**

CODE:-

```python
numbers = [10, 20, 10, 30, 20, 40]

unique_numbers = list(set(numbers))

print("List after removing duplicates:", unique_numbers)
```

**Question 9 : Write a function that returns the square of a number.**

CODE:-

```python
def square(num):
    return num * num

result = square(88)

print("Square is:", result)
```

**Question 10 : Write a Python program to count how many times a number appears in a list.**

**Example List**

**[2,3,4,2,5,2]**

CODE:-

```python
numbers = [2, 3, 4, 2, 5, 2]

count = numbers.count(2)

print("The number 2 appears", count, "times")
```
